```python
import os
import sys
from pathlib import Path

try:
    from PIL import Image
except Exception as e:
    print('Pillow not available:', e)
    raise

root = Path.cwd()
logo_path = root / 'logo.png'
print('logo exists:', logo_path.exists())
if logo_path.exists():
    img = Image.open(logo_path).convert('RGBA')
    pixels = img.load()
    w, h = img.size
    for y in range(h):
        for x in range(w):
            r, g, b, a = pixels[x, y]
            if r > 230 and g > 230 and b > 230:
                pixels[x, y] = (255, 255, 255, 0)
    img.save(logo_path)
    print('saved transparent version to', logo_path)
```
